### Imports Selenium's browser control functionality.

```python
from selenium import webdriver
# Imports Selenium's browser control functionality

driver = webdriver.Chrome()
# Creates a Chrome browser controlled by Python

driver.get("https://www.amazon.in")
# Opens Amazon India in the Chrome browser

print(driver.title)
# Gets and displays the title of the webpage
```

```python
print(driver.title)
```

```python
# replace google link with Amazon
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
river = webdriver.Chrome(
    service=Service(ChromeDriverManager().install())
)

driver.get("https://www.amazon.in")

print(driver.title)

driver.quit()
```
